<h1>Delete outliers from the Deliverable Volume column</h1>

```python
Q1 = reliance["Deliverable Volume"].quantile(0.25)
Q3 = reliance["Deliverable Volume"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

reliance_no_outliers = reliance[
    (reliance["Deliverable Volume"] >= lower_limit) &
    (reliance["Deliverable Volume"] <= upper_limit)
]
```

```python
reliance_no_outliers.shape
```

```python
plt.boxplot(reliance_no_outliers['Deliverable Volume'])
plt.show()
```

```python
Q1 = reliance_no_outliers["Deliverable Volume"].quantile(0.25)
Q3 = reliance_no_outliers["Deliverable Volume"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

remaining = reliance_no_outliers[
    (reliance_no_outliers["Deliverable Volume"] < lower) |
    (reliance_no_outliers["Deliverable Volume"] > upper)
]

print("Remaining outliers:", len(remaining))
```

```python
df_clean = reliance_no_outliers.copy()

while True:
    Q1 = df_clean["Deliverable Volume"].quantile(0.25)
    Q3 = df_clean["Deliverable Volume"].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    new_df = df_clean[
        (df_clean["Deliverable Volume"] >= lower) &
        (df_clean["Deliverable Volume"] <= upper)
    ]

    if len(new_df) == len(df_clean):
        break

    df_clean = new_df

print("Final rows:", len(df_clean))
```

```python
while True:
    Q1 = df_clean["Close"].quantile(0.25)
    Q3 = df_clean["Close"].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    temp = df_clean[
        (df_clean["Close"] >= lower) &
        (df_clean["Close"] <= upper)
    ]

    if len(temp) == len(df_clean):
        break

    df_clean = temp

print("Final rows:", len(df_clean))
```

```python
while True:
    Q1 = df_clean["High"].quantile(0.25)
    Q3 = df_clean["High"].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    old_rows = len(df_clean)

    df_clean = df_clean[(df_clean["High"] >= lower) & (df_clean["High"] <= upper)]

    if len(df_clean) == old_rows:
        break

print("Final rows:", len(df_clean))
```
