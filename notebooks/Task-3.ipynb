<h1>Using scikit-learn's</h1>

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("E:\\My Projects\\Stock price prediction\\cleaned_stock_data.csv")
```

```python
X = df.drop(columns=['High'],axis=1)
y = df['High']
```

```python
X
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train,y_test = train_test_split(X,y,test_size=0.3)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

<h1>Implement Gradient Descent by Hand</h1>

```python
import numpy as np

x_close = X_train['Close'].values.astype(float)
y_high = y_train.values.astype(float)
n = len(x_close)

m = 0.0
c = 0.0
lr = 0.01
epochs = 10000

loss_history = []

for i in range(epochs):
    
    curr_y = m*x_close + c
    error = curr_y - y_high
    
    
    dm = (2/n) * np.sum(error * x_close)
    dc = (2/n) * np.sum(error)
    
    m = m - lr*dm
    c = c - lr*dc
    
print(f"M = {m}")
print(f"C = {c}")
```
