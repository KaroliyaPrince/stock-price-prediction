```python
import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("E:\\My Projects\\Stock price prediction\\cleaned_stock_data.csv")
df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
```

```python
liner_model = LinearRegression()
poly_model = LinearRegression()
SVR_model = SVR(kernel='rbf')
```

```python
liner_model.fit
```

```python
X = df.drop(columns=['High'])
y = df['High']
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=45) 
```

```python
liner_model.fit(X_train , y_train)
```

```python
poly_model.fit(poly.fit_transform(X_train) , y_train)
```

```python
SVR_model.fit(X_train,y_train)
```

```python
y_pred_liner = liner_model.predict(X_test)
y_pred_poly = poly_model.predict(poly.fit_transform(X_test))
y_pred_SVR = SVR_model.predict(X_test)
```

<h1>RSS, RMSE, R²</h1>

<h3>For liner model</h3>

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
RSS = np.sum((y_test - y_pred_liner) ** 2)

RMSE = np.sqrt(mean_squared_error(y_test, y_pred_liner))

R2 = r2_score(y_test, y_pred_liner)

print("RSS :", RSS)
print("RMSE:", RMSE)
print("R²  :", R2)
```

<h3>For Poly. model</h3>

```python
RSS = np.sum((y_test - y_pred_poly) ** 2)

RMSE = np.sqrt(mean_squared_error(y_test, y_pred_poly))

R2 = r2_score(y_test, y_pred_poly)

print("RSS :", RSS)
print("RMSE:", RMSE)
print("R²  :", R2)
```

<h3>For SVR</h3>

```python
RSS = np.sum((y_test - y_pred_SVR) ** 2)

RMSE = np.sqrt(mean_squared_error(y_test, y_pred_SVR))

R2 = r2_score(y_test, y_pred_SVR)

print("RSS :", RSS)
print("RMSE:", RMSE)
print("R²  :", R2)
```

<h1>Check Overfitting / Underfitting</h1>

```python
y_pred_train_liner  = liner_model.predict(X_train)
y_pred_train_poly = poly_model.predict(poly.fit_transform(X_train))
y_pred_train_svr = SVR_model.predict(X_train)
```

<h3>For liner</h3>

```python
train_score = r2_score(y_train, y_pred_train_liner)

# Test R²
test_score = r2_score(y_test, y_pred_liner)

print("Train Score:", train_score)
print("Test Score :", test_score)
```

<h3>For poly</h3>

```python
train_score = r2_score(y_train, y_pred_train_poly)

# Test R²
test_score = r2_score(y_test, y_pred_poly)

print("Train Score:", train_score)
print("Test Score :", test_score)
```

<h3>For SVR</h3>

```python
train_score = r2_score(y_train, y_pred_train_svr)

# Test R²
test_score = r2_score(y_test, y_pred_SVR)

print("Train Score:", train_score)
print("Test Score :", test_score)
```

<h1>Compare All Models</h1>

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Polynomial Regression", "SVR"],
    "RSS": [
        np.sum((y_test - y_pred_liner) ** 2),
        np.sum((y_test - y_pred_poly) ** 2),
        np.sum((y_test - y_pred_SVR) ** 2)
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, y_pred_liner)),
        np.sqrt(mean_squared_error(y_test, y_pred_poly)),
        np.sqrt(mean_squared_error(y_test, y_pred_SVR))
    ],
    "R²": [
        r2_score(y_test, y_pred_liner),
        r2_score(y_test, y_pred_poly),
        r2_score(y_test, y_pred_SVR)
    ]
})

results
```
